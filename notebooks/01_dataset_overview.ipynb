# EduAnalytics-KG: Эксплораторный анализ датасета ОРТ (2006–2025)
Пространственно-временной датасет результатов Общереспубликанского тестирования Кыргызской Республики по районам и регионам.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# 1. Загрузка данных
df_dist = pd.read_csv("../data/04_processed/ort_district_panel_2006_2025.csv")
df_reg = pd.read_csv("../data/04_processed/ort_regional_panel_2006_2025.csv")

print(f"Районная панель: {df_dist.shape[0]} строк, {df_dist.shape[1]} колонок")
print(f"Охвачено лет: {sorted(df_dist['year'].unique())}")
print(f"Количество уникальных районов: {df_dist['district_id'].nunique()}")
df_dist.head()
```

```python
# 2. Динамика средних баллов по регионам
plt.figure(figsize=(14, 7))
sns.lineplot(data=df_dist, x="year", y="mean_score", hue="region_name", errorbar=None, marker="o")
plt.title("Динамика среднего балла ОРТ по регионам Кыргызстана (2006–2025 гг.)", fontsize=14)
plt.xlabel("Год", fontsize=12)
plt.ylabel("Средний тестовый балл", fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# 3. ТОП-10 районов с наивысшими средними баллами в 2025 году
top_2025 = df_dist[df_dist["year"] == 2025].sort_values(by="mean_score", ascending=False).head(10)
print(top_2025[["district_name", "region_name", "mean_score", "pass_rate_pct", "participants_count"]])
```
